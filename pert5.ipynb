```python
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
import matplotlib.pyplot as plt
```

## Contoh 1

```python
df=[['Pena', 'Roti', 'Mentega'],
    ['Roti', 'Mentega', 'Telur', 'Susu'],
    ['Buncis', 'Telur', 'Susu'],
    ['Roti', 'Mentega'],
    ['Roti', 'Mentega', 'Kecap', 'Telur', 'Susu']]
```

```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(df).transform(df)
```

```python
df_apriori = pd.DataFrame(te_ary, columns=te.columns_)
df_apriori
```

```python
from mlxtend.frequent_patterns import apriori

frequent_itemsets=apriori(df_apriori, min_support=0.4, use_colnames=True)
frequent_itemsets
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)

rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```

```python
rules
```

## Contoh 2

```python
df=pd.read_excel('Data/apriori1.xlsx', header=None)
df.head()
```

```python
records = []  
for i in range(df.shape[0]):  
    records.append([str(df.values[i, j]) for j in range(df.shape[1]) if pd.notna(df.values[i, j])])
print(records)
```

```python
from mlxtend.preprocessing import TransactionEncoder
te = TransactionEncoder()
te_ary = te.fit(records).transform(records)
```

```python
df_apriori = pd.DataFrame(te_ary, columns=te.columns_)
df_apriori
```

```python
from mlxtend.frequent_patterns import apriori

apriori(df_apriori, min_support=0.5, use_colnames=True)
```

```python
frequent_itemsets = apriori(df_apriori, min_support=0.5, use_colnames=True)
frequent_itemsets['length'] = frequent_itemsets['itemsets'].apply(lambda x: len(x))
frequent_itemsets
```

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.5)
rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
```
