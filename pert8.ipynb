# Hierarchical Clustering

```python
import warnings 
warnings.filterwarnings('ignore')
```

```python
import numpy as np
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
from scipy.spatial.distance import pdist
```

```python
data=np.array([
    [1,2],
    [2,3],
    [3,6],
    [5,8],
    [8,8]
])
```

```python
# Menghitung matriks jarak menggunakan jarak Manhattan
distance_matrix=pdist(data, metric='cityblock')
print(f'Matriks Jarak:\n{distance_matrix}')

# Metode Linkage yang digunakan
linkage_methods=['single', 'complete', 'average']
linkage_titles=['Single Linkage', 'Complete Linkage', 'Average Linkage']

# Visualisasi dendrogram untuk setiap metode linkage
plt.figure(figsize=(15,5))

for i, method in enumerate(linkage_methods):
    plt.subplot(1, 3, i+1)
    linked=sch.linkage(distance_matrix, method=method)
    sch.dendrogram(linked, labels=['A', 'B', 'C', 'D', 'E'])
    plt.title(f'Dendrogram - {linkage_titles[i]}')
    
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram
import matplotlib.pyplot as plt
```

```python
kota = ['Surabaya', 'Malang', 'Kediri', 'Madiun', 'Jember']
koordinat=np.array([
    [-7.2504, 112.7688],
    [-7.9666, 112.6326],
    [-7.8162, 112.0111],
    [-7.6298, 111.5233],
    [-8.1724, 113.6879]
])
```

```python
manhattan_dist=pdist(koordinat, metric='cityblock')

distance_matrix=pd.DataFrame(squareform(manhattan_dist), index=kota, columns=kota)
distance_matrix
```

```python
methods = ['single', 'complete', 'average']
linkage_titles = ['Single Linkage', 'Complete Linkage', 'Average Linkage']

plt.figure(figsize=(18, 5)) 

for i, method in enumerate(methods):
    plt.subplot(1, 3, i + 1)
    linked = sch.linkage(distance_matrix, method=method)
    sch.dendrogram(linked, labels=kota)
    plt.title(f'Dendrogram - {linkage_titles[i]}')
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
